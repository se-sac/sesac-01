```python
# 필요한 라이브러리 임포트
import torch
from torch import nn
from torch.utils.data import DataLoader

# 모델 불러오는 함수 (예시)
def get_maskrcnn_efficientb2b_model(pretrained=True):
    # 실제 모델 로드 함수로 교체하세요.
    # 여기서는 PyTorch torchvision의 MaskRCNN에 EfficientNetB2 backbone을 사용하는 경우 예시입니다.
    from torchvision.models.detection import maskrcnn_resnet50_fpn
    model = maskrcnn_resnet50_fpn(pretrained=pretrained)
    return model

# 1. 모델 로드
model = get_maskrcnn_efficientb2b_model(pretrained=True)

# 2. 백본 구조 확인 (이름 출력)
print("Backbone named_modules:")
for name, module in model.backbone.named_modules():
    print(name)

# 3. 앞부분 freeze (예시로 backbone의 첫 2개 블록 freeze)
freeze_layers = ['body.layer1', 'body.layer2']  # 실제 이름은 backbone 구조에 따라 다름

for name, param in model.backbone.named_parameters():
    if any(fl in name for fl in freeze_layers):
        param.requires_grad = False

# 4. 학습할 파라미터만 필터링해 옵티마이저 생성
optimizer = torch.optim.SGD(filter(lambda p: p.requires_grad, model.parameters()), lr=0.001, momentum=0.9)

# 5. 학습용 데이터 로더 예시 (사용자 데이터셋으로 교체 필요)
# from your_dataset import CustomDataset
# train_dataset = CustomDataset(...)
# train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True)

# 6. 간단한 학습 루프 예시 (사용자 데이터셋으로 교체)
num_epochs = 5
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

for epoch in range(num_epochs):
    model.train()
    print(f"Epoch {epoch+1}/{num_epochs}")
    # for images, targets in train_loader:
    #     images = [img.to(device) for img in images]
    #     targets = [{k: v.to(device) for k, v in t.items()} for t in targets]
    #     
    #     optimizer.zero_grad()
    #     loss_dict = model(images, targets)
    #     losses = sum(loss for loss in loss_dict.values())
    #     losses.backward()
    #     optimizer.step()
    #
    #     print(f"Loss: {losses.item()}")
    print("Training loop here... (데이터 로더를 연결하세요)")

# 7. freeze 된 파라미터 확인
print("\nFreeze 상태 확인:")
for name, param in model.named_parameters():
    print(f"{name}: requires_grad={param.requires_grad}")
```
